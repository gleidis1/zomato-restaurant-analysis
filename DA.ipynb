```python
import pandas as pd

df = pd.read_csv(r"C:\Users\User\Downloads\Zomato Restaurant Dataset.csv")

df.head()
```

```python

print("Dataset shape:", df.shape)
print("Column names:", df.columns.tolist())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Keep the columns we need
cuisine_df = df[["City", "Cuisines"]].dropna().copy()

# Split multiple cuisines into separate rows
cuisine_df["Cuisines"] = cuisine_df["Cuisines"].str.split(", ")

# Create one row per city-cuisine combination
cuisine_df = cuisine_df.explode("Cuisines")

cuisine_df.head()
```

```python
top_cities = df["City"].value_counts().head(10).index

top_cuisines = (cuisine_df["Cuisines"]
    .value_counts()
    .head(10)
    .index
)

print("Top cities:", list(top_cities))
print("Top cuisines:", list(top_cuisines))
```

```python
# Filter to the top cities and cuisines
filtered = cuisine_df[
    cuisine_df["City"].isin(top_cities) &
    cuisine_df["Cuisines"].isin(top_cuisines)
]

# Count restaurants for each city-cuisine combination
cuisine_pivot = pd.crosstab(
    filtered["City"],
    filtered["Cuisines"]
)

# Plot the heatmap
plt.figure(figsize=(14, 8))

sns.heatmap(
    cuisine_pivot,
    annot=True,
    fmt="d",
    cmap="YlOrRd",
    linewidths=0.5
)

plt.title("Cuisine Distribution Across the Top 10 Cities")
plt.xlabel("Cuisine Type")
plt.ylabel("City")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Split restaurants with multiple cuisines into separate rows
cuisine_df = df[["City", "Cuisines"]].dropna().copy()
cuisine_df["Cuisines"] = cuisine_df["Cuisines"].str.split(", ")
cuisine_df = cuisine_df.explode("Cuisines")

# Select the top 8 cities and top 8 cuisines
top_cities = df["City"].value_counts().head(8).index
top_cuisines = cuisine_df["Cuisines"].value_counts().head(8).index

plot_df = cuisine_df[
    cuisine_df["City"].isin(top_cities) &
    cuisine_df["Cuisines"].isin(top_cuisines)
]

# Count each cuisine in each city
counts = (
    plot_df.groupby(["City", "Cuisines"])
    .size()
    .reset_index(name="Restaurants")
)

# Create colorful grouped bar chart
sns.set_theme(style="whitegrid")
plt.figure(figsize=(16, 8))

sns.barplot(
    data=counts,
    x="City",
    y="Restaurants",
    hue="Cuisines",
    palette="Spectral"
)

plt.title("Which Cuisines Are Where? — Zomato Restaurant Dataset",
          fontsize=18, fontweight="bold", pad=20)
plt.xlabel("City", fontsize=12)
plt.ylabel("Number of Restaurant Listings", fontsize=12)
plt.xticks(rotation=35, ha="right")
plt.legend(title="Cuisine", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```
